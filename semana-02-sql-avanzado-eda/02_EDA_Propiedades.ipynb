en filas, asociando cada columna con su cantidad
de valores nulos.

Posteriormente, se calculó el porcentaje de
valores faltantes respecto al total de registros.

La función ROUND() permitió presentar los
porcentajes redondeados a dos decimales.

### Filtrado de columnas críticas

Se aplicó la condición:

`WHERE porcentaje_nulos > 50`

Esta condición permitió seleccionar únicamente
las columnas con más del 50 % de valores nulos.

Finalmente, se utilizó ORDER BY para ordenar

## 5. Conclusión

El ejercicio permitió identificar cinco

## 6. Conceptos aprendidos

- WITH: definición de múltiples CTEs.
- COUNT(*): conteo total de registros.
- COUNT(columna): conteo de valores no nulos.
- STACK(): transformación de columnas en filas.
- ROUND(): redondeo de porcentajes.
- WHERE: filtrado de resultados.
- ORDER BY: ordenamiento descendente.
- AS: asignación de alias.
- Identificación de columnas con alta
  proporción de valores nulos.
- Evaluación de completitud de datos.

**Estado:** Ejercicio completado, ejecutado
y documentado con resultados reales.


```sql

WITH conteo_nulos AS (
    SELECT
        COUNT(*) AS total_registros,

        COUNT(*) - COUNT(precio) AS nulos_precio,
        COUNT(*) - COUNT(expensas) AS nulos_expensas,
        COUNT(*) - COUNT(tipo_de_operacion) AS nulos_tipo_operacion,
        COUNT(*) - COUNT(moneda) AS nulos_moneda,
        COUNT(*) - COUNT(ambientes) AS nulos_ambientes,

        COUNT(*) - COUNT(metros_cuadrados_totales)
            AS nulos_metros_totales,

        COUNT(*) - COUNT(metros_cuadrados_cubiertos)
            AS nulos_metros_cubiertos,

        COUNT(*) - COUNT(orientacion_cardinal)
            AS nulos_orientacion,

        COUNT(*) - COUNT(piso) AS nulos_piso,
        COUNT(*) - COUNT(cochera) AS nulos_cochera,
        COUNT(*) - COUNT(antiguedad) AS nulos_antiguedad,
        COUNT(*) - COUNT(estado) AS nulos_estado,
        COUNT(*) - COUNT(zona) AS nulos_zona

    FROM workspace.bronze.properties_bronze
),

porcentajes_nulos AS (
    SELECT
        columna,
        ROUND(nulos * 100.0 / total_registros, 2)
            AS porcentaje_nulos

    FROM conteo_nulos,
    LATERAL STACK(
        13,
        'precio', nulos_precio,
        'expensas', nulos_expensas,
        'tipo_de_operacion', nulos_tipo_operacion,
        'moneda', nulos_moneda,
        'ambientes', nulos_ambientes,
        'metros_cuadrados_totales', nulos_metros_totales,
        'metros_cuadrados_cubiertos', nulos_metros_cubiertos,
        'orientacion_cardinal', nulos_orientacion,
        'piso', nulos_piso,
        'cochera', nulos_cochera,
        'antiguedad', nulos_antiguedad,
        'estado', nulos_estado,
        'zona', nulos_zona
    ) AS t(columna, nulos)
)

SELECT
    columna,
    porcentaje_nulos
FROM porcentajes_nulos

WHERE porcentaje_nulos > 50
ORDER BY porcentaje_nulos DESC;
```

## Parte 5 — Detección de problemas de calidad

Se evalúan problemas que pueden afectar

### 5.1 — Reporte de calidad de datos

Se construye un reporte para cuantificar
seis problemas de calidad:

- Precios inválidos.
- Superficies totales inválidas.
- Antigüedad igual a 999.
- Cantidad de ambientes inválida.
- Valores faltantes o vacíos en moneda.
- Valores faltantes o vacíos en tipo de operación.

Se utilizan CTEs, CASE WHEN y funciones
de agregación para obtener las cantidades
y porcentajes de registros afectados.

**Interpretación:**

El reporte permitirá priorizar los problemas
que requieren atención durante la limpieza.

Una misma propiedad puede presentar más
de un problema. Por tanto, los porcentajes
no deben sumarse como si representaran
categorías independientes.

El reporte es un diagnóstico inicial,
no una validación exhaustiva de calidad.


```sql

WITH datos AS (
    SELECT
        TRY_CAST(precio AS DOUBLE) AS precio_num,
        TRY_CAST(metros_cuadrados_totales AS DOUBLE)
            AS metros_num,
        TRY_CAST(antiguedad AS DOUBLE)
            AS antiguedad_num,
        TRY_CAST(ambientes AS DOUBLE)
            AS ambientes_num,
        moneda,
        tipo_de_operacion
    FROM workspace.bronze.properties_bronze
),

reporte AS (
    SELECT
        COUNT(*) AS total_registros,

        SUM(CASE
            WHEN precio_num IS NULL OR precio_num <= 0
            THEN 1 ELSE 0
        END) AS precios_invalidos,

        SUM(CASE
            WHEN metros_num IS NULL OR metros_num <= 0
            THEN 1 ELSE 0
        END) AS metros_invalidos,

        SUM(CASE
            WHEN antiguedad_num = 999
            THEN 1 ELSE 0
        END) AS antiguedad_placeholder,

        SUM(CASE
            WHEN ambientes_num IS NULL OR ambientes_num <= 0
            THEN 1 ELSE 0
        END) AS ambientes_invalidos,

        SUM(CASE
            WHEN moneda IS NULL OR TRIM(moneda) = ''
            THEN 1 ELSE 0
        END) AS moneda_vacia,

        SUM(CASE
            WHEN tipo_de_operacion IS NULL
              OR TRIM(tipo_de_operacion) = ''
              OR LOWER(TRIM(tipo_de_operacion)) = 'null'
            THEN 1 ELSE 0
        END) AS operacion_vacia

    FROM datos
),

problemas AS (
    SELECT
        total_registros,
        problema,
        cantidad
    FROM reporte,
    LATERAL STACK(
        6,
        'Precios invalidos', precios_invalidos,
        'Metros invalidos', metros_invalidos,
        'Antiguedad 999', antiguedad_placeholder,
        'Ambientes invalidos', ambientes_invalidos,
        'Moneda vacia', moneda_vacia,
        'Operacion vacia', operacion_vacia
    ) AS t(problema, cantidad)
)

SELECT
    problema,
    total_registros,
    cantidad,
    ROUND(cantidad * 100.0 / total_registros, 2)
        AS porcentaje
FROM problemas
ORDER BY cantidad DESC;
```

### 5.2 — Detección de registros duplicados

### 5.4 — Detección de precios extremos

Se identifican propiedades cuyos precios
se encuentran fuera del intervalo definido
por los percentiles 1 y 99 de cada moneda.

Se utiliza PERCENTILE_APPROX() para calcular
los límites y una CTE para filtrar los
registros fuera de ese rango.

**Interpretación:**

Los precios identificados son valores extremos
según el criterio estadístico utilizado,
pero no necesariamente errores.

Por ejemplo, una propiedad de lujo puede
tener un precio muy superior al promedio
sin que el registro sea incorrecto.

También debe considerarse que los percentiles
se calcularon por moneda, sin separar
ventas y alquileres.

Por tanto, antes de excluir un registro,
será necesario revisar su contexto.


```sql

WITH precios_validos AS (
    SELECT
        id,
        url,
        moneda,
        tipo_de_operacion,
        TRY_CAST(precio AS DOUBLE) AS precio_num
    FROM workspace.bronze.properties_bronze
),

percentiles AS (
    SELECT
        moneda,
        PERCENTILE_APPROX(precio_num, 0.01)
            AS percentil_1,
        PERCENTILE_APPROX(precio_num, 0.99)
            AS percentil_99
    FROM precios_validos
    WHERE precio_num > 0
      AND moneda IS NOT NULL
    GROUP BY moneda
),

outliers AS (
    SELECT
        p.id,
        p.url,
        p.moneda,
        p.tipo_de_operacion,
        p.precio_num,
        q.percentil_1,
        q.percentil_99
    FROM precios_validos p
    INNER JOIN percentiles q
        ON p.moneda = q.moneda
    WHERE p.precio_num > 0
      AND (
          p.precio_num < q.percentil_1
          OR p.precio_num > q.percentil_99
      )
)

SELECT *
FROM outliers
ORDER BY moneda, precio_num DESC;
```

### 6.1 — Ranking de zonas por precio promedio

Se calcula el precio promedio de las propiedades
en venta, expresadas en USD, para cada zona.

Se consideran únicamente precios numéricos
mayores que cero.

Mediante ROW_NUMBER() se asigna un ranking
ordenado desde la zona con el precio promedio
más alto hasta la de menor promedio.

**Consideración:** Las zonas con pocas propiedades
pueden presentar promedios poco representativos.
Por ello, se incluye la cantidad de registros
utilizados para calcular cada promedio.


```sql

WITH propiedades_validas AS (
    SELECT
        zona,
        TRY_CAST(precio AS DOUBLE) AS precio_num

    FROM workspace.bronze.properties_bronze

    WHERE UPPER(TRIM(moneda)) = 'USD'
      AND LOWER(TRIM(tipo_de_operacion)) = 'venta'
),

promedio_por_zona AS (
    SELECT
        zona,
        COUNT(*) AS cantidad_propiedades,
        ROUND(AVG(precio_num), 2) AS precio_promedio

    FROM propiedades_validas

    WHERE precio_num > 0
      AND zona IS NOT NULL

    GROUP BY zona
)

SELECT
    ROW_NUMBER() OVER (
        ORDER BY precio_promedio DESC, zona
    ) AS ranking,

    zona,
    cantidad_propiedades,
    precio_promedio

FROM promedio_por_zona

ORDER BY ranking;
```

### 6.2 — Comparación con el promedio general

Se compara el precio promedio de cada zona
con el promedio general de las propiedades
en venta expresadas en USD.

Se utiliza AVG() OVER() para obtener
el promedio general sin eliminar el detalle
de los registros.

Posteriormente, se calculan:

- Precio promedio por zona.
- Precio promedio general.
- Diferencia absoluta.
- Diferencia porcentual.

Una diferencia positiva indica que la zona
presenta un precio promedio superior al general.

Una diferencia negativa indica que su promedio
está por debajo del promedio general.

La comparación considera únicamente propiedades
con precios válidos y zonas no nulas.


```sql

-- Ejercicio 6.2: Comparacion con promedio general

WITH propiedades_validas AS (
    SELECT
        zona,
        TRY_CAST(precio AS DOUBLE) AS precio_num

    FROM workspace.bronze.properties_bronze

    WHERE UPPER(TRIM(moneda)) = 'USD'
      AND LOWER(TRIM(tipo_de_operacion)) = 'venta'
),

precios_comparados AS (
    SELECT
        zona,
        precio_num,

        AVG(precio_num) OVER ()
            AS promedio_general

    FROM propiedades_validas

    WHERE precio_num > 0
      AND zona IS NOT NULL
),

promedios_zona AS (
    SELECT
        zona,
        COUNT(*) AS cantidad_propiedades,
        AVG(precio_num) AS promedio_zona,
        MAX(promedio_general) AS promedio_general

    FROM precios_comparados

    GROUP BY zona
)

SELECT
    zona,
    cantidad_propiedades,

    ROUND(promedio_zona, 2) AS promedio_zona,
    ROUND(promedio_general, 2) AS promedio_general,

    ROUND(
        promedio_zona - promedio_general,
        2
    ) AS diferencia_absoluta,

    ROUND(
        (promedio_zona - promedio_general)
        * 100.0 / promedio_general,
        2
    ) AS diferencia_porcentual

FROM promedios_zona

ORDER BY promedio_zona DESC;
```

### 6.3 — Análisis temporal de propiedades

Se analiza la cantidad de propiedades y su
precio promedio agrupando los registros por mes.

La columna fecha se convierte a TIMESTAMP
y posteriormente se utiliza DATE_TRUNC()
para agrupar las observaciones mensualmente.

Se consideran propiedades en venta, expresadas
en USD, con fechas y precios válidos.

**Consideración:**

Las variaciones del precio promedio pueden
estar relacionadas con cambios en la composición
de las propiedades registradas.

Por tanto, una variación mensual no demuestra

```sql

-- Ejercicio 6.3: Analisis temporal

WITH propiedades_fechadas AS (
    SELECT
        TRY_CAST(fecha AS TIMESTAMP) AS fecha_valida,
        TRY_CAST(precio AS DOUBLE) AS precio_num

    FROM workspace.bronze.properties_bronze

    WHERE UPPER(TRIM(moneda)) = 'USD'
      AND LOWER(TRIM(tipo_de_operacion)) = 'venta'
),

propiedades_validas AS (
    SELECT
        DATE_TRUNC('MONTH', fecha_valida) AS mes,
        precio_num

    FROM propiedades_fechadas

    WHERE fecha_valida IS NOT NULL
      AND precio_num > 0
)

SELECT
    mes,
    COUNT(*) AS cantidad_propiedades,
    ROUND(AVG(precio_num), 2) AS precio_promedio

FROM propiedades_validas

GROUP BY mes

ORDER BY mes;
```

```sql

-- Ejercicio 7.1: Resumen general de calidad

WITH datos AS (
    SELECT
        TRY_CAST(precio AS DOUBLE) AS precio_num,

        TRY_CAST(
            metros_cuadrados_totales AS DOUBLE
        ) AS metros_num,

        TRY_CAST(ambientes AS DOUBLE) AS ambientes_num,

        TRY_CAST(antiguedad AS DOUBLE)
            AS antiguedad_num,

        moneda,
        tipo_de_operacion

    FROM workspace.bronze.properties_bronze
),

validacion AS (
    SELECT
        CASE
            WHEN precio_num > 0
             AND metros_num > 0
             AND ambientes_num > 0

             AND moneda IS NOT NULL
             AND TRIM(moneda) <> ''

             AND tipo_de_operacion IS NOT NULL
             AND TRIM(tipo_de_operacion) <> ''
             AND LOWER(TRIM(tipo_de_operacion)) <> 'null'

             AND (
                 antiguedad_num IS NULL
                 OR antiguedad_num <> 999
             )

            THEN 1
            ELSE 0
        END AS es_valido

    FROM datos
)

SELECT
    COUNT(*) AS total_registros,

    SUM(es_valido) AS registros_validos,

    COUNT(*) - SUM(es_valido)
        AS registros_invalidos,

    ROUND(
        SUM(es_valido) * 100.0 / COUNT(*),
        2
    ) AS porcentaje_validos,

    ROUND(
        (COUNT(*) - SUM(es_valido))
        * 100.0 / COUNT(*),
        2
    ) AS porcentaje_invalidos

FROM validacion;
```
